```python
import pandas as pd

df = pd.read_csv('Downloads/customer_shopping_behavior.csv')
```

```python
#filling missing values in review rating with median of each category
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#converting to snake case 
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
x=df.promo_code_used==df.discount_applied
x.sum()
```

```python
#dropping promo_code column as it is same as discount applied
df=df.drop("promo_code_used",axis=1)
df.columns
```

```python
#transforming frequency of purchase columns to numerical value
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df.head()
```

```python
df["age"].max()
```

```python
df["age"].min()
```

```python
labels = ['Young', 'Adult', 'Middle-aged', 'Senior']
df["age_buckets"]=pd.qcut(df["age"],q=4,labels=labels)
df.head()
```
